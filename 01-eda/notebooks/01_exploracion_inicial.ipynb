```python
df = pd.read_csv(
    '../data/raw/exportaciones_por_pais.csv',
    encoding='latin-1',
    skiprows=[0, 2]
)

df = df.rename({'DESCRIPCIÓN':'periodo'}, axis=1)
```

```python
# Dic de parseo fechas
meses = {
    'ENE':'01', 'FEB':'02',
    'MAR':'03', 'ABR':'04',
    'MAY':'05', 'JUN':'06',
    'JUL':'07', 'AGO':'08',
    'SEP':'09', 'OCT':'10',
    'NOV':'11', 'DIC':'12'
}
```

```python
def parear_periodo(periodo):
    mes_abr = periodo[:3]
    anio = periodo[3:]
    return f'{anio}-{meses[mes_abr]}-01'

df['periodo'] = pd.to_datetime(df['periodo'].apply(parear_periodo))
```

```python
columnas_totales = [c for c in df.columns if 'Total' in c]
columnas_paises = [c for c in df.columns if c not in columnas_totales and c != 'periodo']

print(columnas_totales)
print(columnas_paises)
```

```python
df_paises = df.melt(
    id_vars='periodo',
    value_vars=columnas_paises,
    var_name='destino',
    value_name='exportaciones_musd'
)

df_totales = df.melt(
    id_vars='periodo',
    value_vars=columnas_totales,
    var_name='region',
    value_name='exportaciones_musd'
)

df_paises.head()
```

```python
partes = df_paises['destino'].str.split('. ', expand=True, regex=False)
partes.columns = ['continente', 'subregion', 'pais']
partes.head()
```

```python
df_paises = pd.concat([df_paises, partes], axis=1)
df_paises[df_paises['continente'] == 'Europa']['pais'].unique()
```

```python
df_paises.loc[df_paises['pais'].isna(), 'pais'] = (
    df_paises.loc[df_paises['pais'].isna(), 'subregion']
)
df_paises[df_paises['continente'] == 'Europa'].head()
```

```python
mask_dos_niveles = df_paises['subregion'] == df_paises['pais']
df_paises.loc[mask_dos_niveles, 'subregion'] = None
df_paises[['continente', 'subregion', 'pais']].drop_duplicates()
```
