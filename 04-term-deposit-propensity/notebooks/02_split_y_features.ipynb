```python
import sys; sys.path.append("..")
import pandas as pd
from src.bq_utils import run

df = run(
    "SELECT * FROM " \
    "`TU_PROYECTO.TU_DATASET.bank_additional_full`" \
    "ORDER BY row_id"
)
df["y_bin"] = (df["y"] == "yes").astype(int)
df.shape
```

```python
n = len(df)

cortes = [
    (0.6, 0.2, 0.2),
    (0.5, 0.25, 0.25),
    (0.7, 0.15, 0.15)
]

for fr in cortes:
    a, b = int(n * fr[0]), int(n * (fr[0] + fr[1]))
    tr, va, te = df.y_bin[:a], df.y_bin[a:b], df.y_bin[b:]
    print(fr, [round(x.mean(), 3) for x in (tr, va, te)])
```
